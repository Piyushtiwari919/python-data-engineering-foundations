```python
import time

import numpy as np
```

### List v/s Numpy Array

```python
start_time = time.perf_counter()

list_py = [i*2 for i in range(10000000)]

end_time = time.perf_counter()

print(f"List creation takes {(end_time-start_time):.4f} seconds")
```

```python
start_time = time.perf_counter()

np_array = np.arange(10000000)*2

end_time = time.perf_counter()

print(f"Numpy Array creation takes {(end_time-start_time):.4f} seconds")
```

#### Generating Pattern
